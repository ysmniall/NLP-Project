# Normalization

```python
def normalize_text(text):
    api_url = "https://api.pars.io/v1/nlp/normalize"
    headers = {"Content-Type": "application/json"}
    payload = {"text": text}

    try:
        response = requests.post(api_url, json = payload, headers = headers)
        if response.status_code == 200:
            return response.json().get("result", text)
        else:
            return text
    except:
        return text
```

# Using an API

```python
for province in provinces:
  province_recipes_file = os.listdir("./" + province + "Modified/")
  # print(province_recipes_file)
  with open("./" + province + "Modified/" + province_recipes_file[0], "r", encoding = "utf-8") as f:
      original_data = json.load(f)
      normalized_text = normalize_text(original_data)
      # print(normalized_text[0])
      normalization_file_name = province[0:len(province)-1] + "_normalized.json"
      normalized_file_path = "./" + province + "Modified/" + normalization_file_name
      with open(normalized_file_path, 'w', encoding = 'utf-8') as f2:
          json.dump(normalized_text, f2, ensure_ascii = False, indent = 4)
          f2.close()
      print(province[0:len(province)-1] + " Normalized file created successfully!")
      f.close()
```

# Using DadmaTools

```python
dadma_normalizer = Nm(
    full_cleaning=False,
    unify_chars=True,
    refine_punc_spacing=True,
    remove_extra_space=True,
    remove_puncs=False,
    remove_html=False,
    remove_stop_word=False,
    replace_email_with="<EMAIL>",
    replace_number_with=None,
    replace_url_with="",
    replace_mobile_number_with=None,
    replace_emoji_with=None,
    replace_home_number_with=None
)


for province in provinces:
  normalized_text = []
  province_recipes_file = os.listdir("./" + province + "Modified/")
  # print(province_recipes_file)
  with open("./" + province + "Modified/" + province_recipes_file[0], "r", encoding = "utf-8") as f:
      original_data = json.load(f)
      for recipe_json in tqdm.notebook.tqdm(original_data):
          for item in recipe_json.items():
              if item[0] == 'instructions':
                  normalized_text.append(dadma_normalizer.normalize(item[1][0]))
      normalization_file_name = province[0:len(province)-1] + "_normalized_with_DadmaTools.json"
      normalized_file_path = "./" + province + "Modified/" + normalization_file_name
      with open(normalized_file_path, 'w', encoding = 'utf-8') as f2:
          json.dump(normalized_text, f2, ensure_ascii = False, indent = 4)
          f2.close()
      print(province[0:len(province)-1] + " Normalized file using dadmatools created successfully!")
      f.close()
```

# Normalization, Sentence Tokenization and Word Tokenization using Hazm (The most complete one)




```python
# hazm_normalizer = Normalizer()
__recipes_sentences = []
__recipe_words = []

for province in provinces:
  normalized_text = []
  province_recipes_file = os.listdir("./" + province + "Modified/")
  # print(province_recipes_file)
  with open("./" + province + "Modified/" + province_recipes_file[0], "r", encoding = "utf-8") as f:
      original_data = json.load(f)
      # print(len(original_data))

      for recipe_json in tqdm.notebook.tqdm(original_data):
          for item in recipe_json.items():
              if item[0] == 'instructions':
                  normalized_text.append(dadma_normalizer.normalize(item[1][0]))

      recipes_sentences = [sent_tokenize(x) for x in tqdm.notebook.tqdm(normalized_text)]
      __recipes_sentences += recipes_sentences
      recipe_tokens = [[word_tokenize(sent) for sent in sents] for sents in tqdm.notebook.tqdm(recipes_sentences)]
      __recipe_words += recipe_tokens
      # print(normalized_text[0])
      normalization_file_name = province[0:len(province)-1] + "_normalized_with_Hazm.json"
      sentence_tokenization_file_name = province[0:len(province)-1] + "_normalized_and_tokenized_with_Hazm.json"
      word_tokenization_file_name = province[0:len(province)-1] + "_normalized_and_sentence_and_word_tokenized_with_Hazm.json"
      normalized_file_path = "./" + province + "Modified/" + normalization_file_name
      sentence_tokenization_file_path = "./" + province + "Modified/" + sentence_tokenization_file_name
      word_tokenization_file_path = "./" + province + "Modified/" + word_tokenization_file_name
      with open(normalized_file_path, 'w', encoding = 'utf-8') as f2:
          json.dump(normalized_recipes, f2, ensure_ascii = False, indent = 4)
          f2.close()
      with open(sentence_tokenization_file_path, 'w', encoding = 'utf-8') as f3:
          json.dump(recipes_sentences, f3, ensure_ascii = False, indent = 4)
          f3.close()
      with open(word_tokenization_file_path, 'w', encoding = 'utf-8') as f4:
          json.dump(recipe_tokens, f4, ensure_ascii = False, indent = 4)
          f4.close()
      print(province[0:len(province)-1] + " Normalized and Sentence Tokenized and Word Tokenized files created successfully!")
      f.close()
```

# Frequency Analysis for All Province Data

```python
# nltk.download()
from nltk import FreqDist
import itertools
```

```python
province_tokens = []
dataframe = {}
province_index = 0
for province in provinces:
  province_recipes_file = os.listdir("./" + province + "Modified/")
  word_tokenization_file_name = province[0:len(province)-1] + "_normalized_and_sentence_and_word_tokenized_with_Hazm.json"
  word_tokenization_file_path = "./" + province + "Modified/" + word_tokenization_file_name
  with open(word_tokenization_file_path, 'r', encoding = 'utf-8') as f:
      word_tokenized_data = json.load(f)
      province_tokens.append(itertools.chain(*itertools.chain(*word_tokenized_data)))
      # dataframe[province] = FreqDist(eval(F"province_tokens")).most_common(25)
      dataframe[province] = FreqDist(province_tokens[province_index]).most_common(25)
      province_index += 1
      f.close()

freq_analysis = pd.DataFrame(dataframe)
left_aligned_freq_analysis = freq_analysis.style.set_properties(**{'text-align': 'left'})
print("Word Frequency Analysis using Hazm:")
display(freq_analysis)
print("\n")
```
